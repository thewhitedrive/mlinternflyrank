## 3. Success metric

*One metric you can defend. What number means 'good'?*

Primary metric: Precision@K, with K = 20 and K = 50.

Precision@K measures the fraction of pages in the top-K ranked opportunities that are actually associated with decline. This matches the operational goal because the model will produce a short ranked review queue rather than making a decision about every page.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Build the CTR opportunity proxy

tier_ctr = df.groupby("position_tier")["ctr"].transform("mean")

df["ctr_gap"] = df["ctr"] - tier_ctr

# Bootstrap decline label for evaluation
df["is_declining_label"] = (
    df["trend_direction"] == "down"
).astype(int)

# Opportunity score:
# more negative CTR gap = greater potential opportunity
df["opportunity_score"] = -df["ctr_gap"]

print("CTR gap summary:")
display(df["ctr_gap"].describe())

print("\nDecline label rate:")
print(df["is_declining_label"].mean())
```

```python
def precision_at_k(scores, labels, k):
    top_k = scores.sort_values(ascending=False).head(k).index
    return labels.loc[top_k].mean()

for k in [20, 50]:
    precision = precision_at_k(
        df["opportunity_score"],
        df["is_declining_label"],
        k
    )
    print(f"Precision@{k}: {precision:.3f}")
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule based on a single signal, such as CTR or position, cannot capture all of the variation in content performance. Pages with similar positions can have very different CTRs and decline patterns. Other signals such as impressions, freshness, content type, engagement, and search demand may also contribute to the opportunity.

Therefore, the problem is better suited to a learned scoring approach that can combine multiple signals and produce a ranked list for review. ML does not make the final refresh decision; it provides decision support.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Check how decline rates vary across position tiers

decline_by_position = (
    df.groupby("position_tier")["is_declining_label"]
      .agg(["count", "mean"])
      .sort_values("mean", ascending=False)
)

decline_by_position

```

```python
# CTR varies substantially even within the same position tier

ctr_by_position = (
    df.groupby("position_tier")["ctr"]
      .agg(["count", "mean", "median", "std"])
      .sort_values("mean", ascending=False)
)

ctr_by_position
```

```python
# Compare decline rates across freshness tiers

decline_by_freshness = (
    df.groupby("freshness_tier")["is_declining_label"]
      .agg(["count", "mean"])
      .sort_values("mean", ascending=False)
)

decline_by_freshness
```

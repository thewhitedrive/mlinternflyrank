```python
import pandas as pd
url = "https://raw.githubusercontent.com/thewhitedrive/mlinternflyrank/main/data/raw/content_refresh_anonymized.csv"
df=pd.read_csv(url)
print(" Shape :",df.shape)
print("\nColumns :")
print(df.columns.tolist())
```

```python
df.describe()
```

```python
df.info()
```

```python
# Show columns related to ranking, traffic and engagement

important_words = [
    "position",
    "impression",
    "click",
    "ctr",
    "session",
    "engagement",
    "search",
    "pageview"
]

for col in df.columns:
    if any(word in col.lower() for word in important_words):
        print(col)

```

```python
df[
    [col for col in df.columns
     if any(word in col.lower() for word in
            ["position", "impression", "click", "ctr", "engagement"])]
].describe()
```

```python
print("Number of content items:", len(df))

print("Median 90-day clicks:", df["clicks_90d"].median())

print("Median 90-day impressions:", df["impressions_90d"].median())

print("Median average position:", df["avg_position"].median())
```
